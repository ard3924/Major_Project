```python

x = df[['ACCESS_ALL_DOWNLOADS',
'ACCESS_CACHE_FILESYSTEM',
'ACCESS_FINE_LOCATION',
'ACCESS_NETWORK_STATE',
'ACCESS_SERVICE',
'ACCESS_SHARED_DATA',
'ACCESS_SUPERUSER',
'ACCESS_WIFI_STATE',
'CAMERA',
'CHANGE_CONFIGURATION',
'DELETE_CACHE_FILES',
'READ_ATTACHMENT',
'READ_CONTACTS',
'READ_DATA',
'READ_EXTERNAL_STORAGE',
'READ_GMAIL',
'READ_HISTORY_BOOKMARKS',
'READ_MESSAGES',
'READ_PHONE_STATE',
'READ_SETTINGS',
'READ_SMS',
'RECEIVE_BOOT_COMPLETED',
'RECEIVE_SMS']]
```

```python
x
```

```python
y =df['Label']
```

```python
from imblearn.over_sampling import RandomOverSampler
ro=RandomOverSampler()
x_data,y_data=ro.fit_resample(x,y)
```

```python
from collections import Counter
print("Actual Data:",Counter(y))
```

```python
print("Artificial Data:",Counter(y_data))
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x_data, y_data,
    test_size = 0.25,
    random_state = 42
)
```

```python
 
from sklearn.linear_model import LogisticRegression
 
 
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
```

```python
rfc_model =LogisticRegression()

rfc_model.fit(x_train, y_train)

rfc_xtest_pred = rfc_model.predict(x_test)
rfc_xtrain_pred = rfc_model.predict(x_train)
```

```python
rfc_xtest_accuracy = accuracy_score(y_test, rfc_xtest_pred)
rfc_xtrain_accuracy = accuracy_score(y_train, rfc_xtrain_pred)

print(f"X test Accuracy = {rfc_xtest_accuracy * 100} %")
print(f"X train Accuracy = {rfc_xtrain_accuracy * 100} %")
```

```python
print(classification_report(y_test, rfc_xtest_pred))
```

```python
y_pred = rfc_model.predict(x_test )
y_true=y_test


from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_true,y_pred)
cm
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

f, ax=plt.subplots(figsize=(5,5))
sns.heatmap(cm,annot=True,linewidths=0.5,linecolor="red",fmt=".0f",ax=ax)
plt.xlabel("y_pred")
plt.ylabel("y_true")
plt.show()
```

```python
import pickle

pickle.dump(rfc_model,open('logic_malware.pkl','wb'))
random_malware = pickle.load(open('logic_malware.pkl','rb'))
```
